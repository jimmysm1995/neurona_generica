# Hola Neurona

## Una neurona con una entrada (y un peso)

Establecemos un peso de 0.5 y pedimos al usuario que introduzca un valor de entrada. La salida de la neurona será el producto de la entrada por el peso.

```python
peso = 0.5
entrada = float(input("Introduzca el valor de la entrada: "))
print("La salida de la neurona es ",entrada * peso)
```

## Una neurona con dos entradas (y dos pesos)
Establecemos un peso de 1.5 y 2.5 y pedimos al usuario que introduzca dos valores de entrada. La salida de la neurona será la suma de los productos de las entradas por los pesos.

```python
peso0, peso1 = 1.5, 2.5
entrada0, entrada1  = [float(n) for n in input("Introduzca el valor de la primera entrada y la segunda separada por un espacio: ").split()]
print("La salida de la neurona es ",entrada0 * peso0 + entrada1 * peso1)
```

## Una neurona con tres entradas y bias (sesgo)

```python
peso0, peso1, peso2 = 1, 2, 3
bias = 10
entrada0, entrada1, entrada2  = [float(n) for n in input("Introduzca el valor de las 3 entradas separadas por un espacio: ").split()]
print("La salida de la neurona es ",entrada0 * peso0 + entrada1 * peso1 + entrada2 * peso2 + bias)
```

## Una neurona con n entradas y bias (sesgo)

```python
# Haremos lo mismo, pero usando un bucle para pedir las entradas y los pesos
# y añadiremos un bias (sesgo) a la salida

pesos = [float(n) for n in input("Introduzca los pesos separados por espacios: ").split()]
entradas = [float(n) for n in input("Introduzca las entradas separadas por espacios: ").split()]
bias = float(input("Introduzca el valor del bias: "))

if len(pesos) != len(entradas):
    print("El número de pesos y entradas no coincide")
else:
    print("La salida de la neurona es ",sum([pesos[i] * entradas[i] for i in range(len(pesos))]) + bias)
```
